```python
import pandas as pd
```

```python
customers_df = pd.read_csv('/content/customers-100000.csv')
print('First 10 rows of Customers Data:')
display(customers_df.head(10))
```

## Customers Data Features

*   **Index**: The row number.
*   **Customer Id**: A unique identifier for the customer.
*   **First Name**: The first name of the customer.
*   **Last Name**: The last name of the customer.
*   **Company**: The company the customer is associated with.
*   **City**: The city where the customer is located.
*   **Country**: The country where the customer resides.
*   **Phone 1 / Phone 2**: Contact phone numbers.
*   **Email**: The customer's email address.
*   **Subscription Date**: The date the customer subscribed.
*   **Website**: The customer's personal or company website.

```python
organizations_df = pd.read_csv('/content/organizations-1000.csv')
print('\nFirst 10 rows of Organizations Data:')
display(organizations_df.head(10))
```
